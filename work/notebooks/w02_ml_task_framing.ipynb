## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success metric: F1-score

I will use F1-score as the success metric because the task is a binary classification problem and both false positives and false negatives are important.

F1-score is the harmonic mean of precision and recall. It gives a single measure that balances the ability to correctly identify declining content with the ability to avoid incorrectly flagging non-declining content.

A higher F1-score indicates better performance. The maximum possible F1-score is 1.0, which represents perfect precision and recall, while 0 represents the worst possible score.

For this task, a higher F1-score would mean the model is better at identifying declining content while maintaining a reasonable number of correct predictions.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could identify declining content using a few manually selected conditions, such as setting a threshold for clicks, CTR, or average position.

However, content performance depends on multiple signals at the same time. Different content items can have different content types, traffic levels, engagement patterns, search performance, and other characteristics. These signals may interact with each other, making it difficult to create a single set of fixed if-else rules that works well for all content items.

Machine learning can learn patterns from multiple signals in the dataset and use them together to classify content as declining or not declining.

The benefit is not simply making a prediction. The prediction can help the content/editorial team prioritize which content items should be reviewed first.

A wrong prediction has a cost: a false positive may waste time reviewing content that is not declining, while a false negative may cause genuinely declining content to be missed.

Therefore, ML may provide more flexible decision support than a small number of manually defined rules.
